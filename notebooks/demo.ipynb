```python
import pandas as pd
from kv2026_zvoleni import predict

kandidati = pd.DataFrame([
    dict(poradi=1, mandaty=9, pocet_kandidatek=2, pocet_kandidatu=18, obyvatel=450, vek=52, zena=0, titul=0, starosta=1, typ="M"),
    dict(poradi=3, mandaty=15, pocet_kandidatek=4, pocet_kandidatu=50, obyvatel=2500, vek=38, zena=1, titul=1, starosta=0, typ="S"),
    dict(poradi=25, mandaty=35, pocet_kandidatek=9, pocet_kandidatu=260, obyvatel=60000, vek=29, zena=0, titul=0, starosta=0, typ="K"),
])
kandidati["sance"] = predict(kandidati).round(3)
kandidati
```

Špatný vstup funkce odmítne:

```python
try:
    predict(kandidati.assign(typ="X"))
except ValueError as e:
    print("chyba:", e)
```
